## 04｜把程式存成 worklog.py 檔
**這一步在做什麼**：把下方 Code 格裡的程式，存成一個獨立的檔案 `worklog.py`。

> **為什麼要做這一步？** 寫在 Notebook 格子裡的程式，只活在這份 Notebook 裡。把它存成 .py 檔，就像把草稿謄成正式文件：可以單獨交出去，也可以單獨執行。

**怎麼做**：在下方 Code 格按 Shift + Enter。

- 第一行 `%%writefile worklog.py` 是 Jupyter 的存檔指令，意思是「把這一格的內容寫進 worklog.py」。這一行本身不會被寫進檔案。
- **這一步只存檔，不會計算**，所以還看不到 75 分鐘。就像書狀存檔了，不等於已經遞狀。

**做對了會看到**：`Writing worklog.py`（第一次建立）或 `Overwriting worklog.py`（覆蓋舊檔）。

> **注意**：每執行一次，就會覆蓋同名的舊檔。想保留舊版，請先另存一份。

````python
%%writefile worklog.py
"""案件工時紀錄器：虛構教學資料，單位為整數分鐘。"""


def total_minutes(minutes):
    """加總非負整數分鐘；空清單回傳 0。"""
    total = 0
    for value in minutes:  # 一筆一筆檢查
        # True／False 在 Python 裡會被當成 1／0，所以要另外擋下來
        if isinstance(value, bool) or not isinstance(value, int):
            raise TypeError("分鐘必須是整數，不能是文字或小數。")
        if value < 0:
            raise ValueError("分鐘不能是負數。")
        total += value  # 通過檢查，才加進總數
    return total


def main():
    # 學生可修改這一行的活動名稱及分鐘。
    records = [("會議", 45), ("整理資料", 30)]
    # 只取出每筆的分鐘數，交給 total_minutes 加總
    total = total_minutes([minutes for activity, minutes in records])
    for activity, minutes in records:
        print(f"{activity}：{minutes} 分鐘")
    hours, remaining = divmod(total, 60)  # 換算成幾小時又幾分鐘
    print(f"總計：{total} 分鐘（{hours} 小時 {remaining} 分鐘）")


# 直接執行這個檔案時（python worklog.py），才啟動 main()
if __name__ == "__main__":
    main()
````

### 讀懂這段程式（不用背，看得懂意思就好）

| 程式 | 白話意思 |
|---|---|
| `def total_minutes(minutes):` | 做一個「加總工時」的小工具：交給它一串分鐘數，它算出總和 |
| `for value in minutes:` | 一筆一筆檢查 |
| `isinstance(...)` 那一行 | 不是整數就退件（文字、小數都不行） |
| `if value < 0:` | 負數也退件 |
| `raise TypeError(...)`、`raise ValueError(...)` | 退件，並顯示括號裡的錯誤原因 |
| `total += value` | 檢查通過，加進總數 |
| `def main():` | 主程式：這個工具實際要做的事 |
| `records = [("會議", 45), ("整理資料", 30)]` | 活動資料，每筆是「（名稱, 分鐘）」。**這是唯一需要修改的地方** |
| `divmod(total, 60)` | 總分鐘除以 60，得到「幾小時」和「剩幾分鐘」 |
| `print(...)` | 把結果顯示出來 |
| `if __name__ == "__main__":` | 「直接執行這個檔案時，就啟動 main()」。先當成固定寫法即可 |

**為什麼連 True／False 也要擋？** 在 Python 裡，True 會被當成 1、False 會被當成 0。如果不擋，有人誤填 True，程式會默默算成 1 分鐘，也不會報錯。這種「看起來正常、其實算錯」的情況最危險，所以要明確拒絕。

> **為什麼要做這一步？** Notebook 會記得之前執行過的內容。有時程式看起來正常，其實是借用了 Notebook 的記憶。這一格讓一個什麼都不記得的全新 Python 來執行，就像請沒看過草稿的人讀正式文件，確認文件本身就讀得懂。

**做對了會看到**：一樣是 75 分鐘，最後一行是 `結束代碼： 0 （0 代表成功）`。結束代碼是程式結束時回報的狀態：0 表示正常結束，其他數字表示出錯。

````python
# 老師示範：另外啟動一個全新的 Python 執行 worklog.py（學生看結果即可）
import subprocess
import sys
from pathlib import Path

script = Path("worklog.py").resolve()
if not script.is_file():
    print("請先執行第 04 步存檔的那一格。")
else:
    result = subprocess.run(
        [sys.executable, "-X", "utf8", str(script)],
        capture_output=True, text=True, encoding="utf-8", timeout=10
    )
    print(result.stdout, end="")
    if result.stderr:
        print(result.stderr)
    print("結束代碼：", result.returncode, "（0 代表成功）")
    if result.returncode != 0:
        raise RuntimeError("執行失敗，請保留錯誤訊息。")
````

### 老師示範②：用 8 組已知答案檢查
**這一步在做什麼**：自動測試 8 組資料，其中 4 組應該算出正確答案，4 組不合理的資料應該被拒絕。**學生看結果就好。**

> **為什麼要做這一步？** 只試一組資料，可能剛好碰對。審查契約不會只看一條，測試也要涵蓋一般情況、特殊情況，以及不合理的輸入。好的工具不只要算對，還要擋住錯的資料。

| 測試資料 | 預期結果 | 為什麼測這組 |
|---|---|---|
| 45、30 | 75 | 跟手算答案對照 |
| 空清單（沒有任何活動） | 0 | 特殊情況：什麼都沒填 |
| 0 | 0 | 特殊情況：0 分鐘是合法的 |
| 45、30、20 | 95 | 多一筆也要算對 |
| -1 | 拒絕 | 工時不可能是負數 |
| 文字 "30" | 拒絕 | 文字不是數字，不能相加 |
| 1.5 | 拒絕 | 本工具只收整數分鐘 |
| True | 拒絕 | 避免被默默當成 1 分鐘 |

**做對了會看到**：`通過 8 組檢查：4 組正確資料，以及 4 組不合法輸入。` 有任何一組沒通過，會出現錯誤訊息並停下來。

````python
# 老師示範：用 8 組已知答案測試 worklog.py 裡的 total_minutes（學生看結果即可）
import importlib.util
from pathlib import Path

script = Path("worklog.py").resolve()
if not script.is_file():
    print("請先執行第 04 步存檔的那一格。")
else:
    spec = importlib.util.spec_from_file_location("lesson_worklog", script)
    module = importlib.util.module_from_spec(spec)
    spec.loader.exec_module(module)
    for values, expected in [([45, 30], 75), ([], 0), ([0], 0), ([45, 30, 20], 95)]:
        assert module.total_minutes(values) == expected
    for values, error in [([-1], ValueError), (["30"], TypeError), ([1.5], TypeError), ([True], TypeError)]:
        try:
            module.total_minutes(values)
        except error:
            pass
        else:
            raise AssertionError(f"這組資料應該被拒絕：{values!r}")
    print("通過 8 組檢查：4 組正確資料，以及 4 組不合法輸入。")
````

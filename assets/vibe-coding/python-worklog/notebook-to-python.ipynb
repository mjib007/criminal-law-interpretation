## 04｜產生 worklog.py
**這格在做什麼**：把以下完整內容寫成檔案。**這一步只存檔，不會計算工時。**

操作位置：**Notebook 的新 Code 格**。`%%writefile worklog.py` 必須在第一行，它是 IPython 指令，不會被寫進產生的 `.py`。

預期看到 `Writing worklog.py` 或 `Overwriting worklog.py`。

> 再執行會覆寫目前資料夾的同名檔，要保留舊版請先另存。AI 答案外側的三個反引號不要貼進 Code 格。本課後續格子只依賴磁碟上的 worklog.py，不依賴前一格的 Python 變數；沒檔案時會提示先產生。

````python
%%writefile worklog.py
"""案件工時紀錄器：虛構教學資料，單位為整數分鐘。"""


def total_minutes(minutes):
    """加總非負整數分鐘；空清單回傳 0。"""
    total = 0
    for value in minutes:
        if isinstance(value, bool) or not isinstance(value, int):
            raise TypeError("分鐘必須是整數，不能是文字或小數。")
        if value < 0:
            raise ValueError("分鐘不能是負數。")
        total += value
    return total


def main():
    # 學生可修改這一行的活動名稱及分鐘。
    records = [("會議", 45), ("整理資料", 30)]
    total = total_minutes([minutes for activity, minutes in records])
    for activity, minutes in records:
        print(f"{activity}：{minutes} 分鐘")
    hours, remaining = divmod(total, 60)
    print(f"總計：{total} 分鐘（{hours} 小時 {remaining} 分鐘）")


if __name__ == "__main__":
    main()
````

### 教師驗證：讓另一個 Python 執行
**這格在做什麼**：用相同 Python 開啟新行程，確認不依賴 Notebook 記住的變數。預期同樣出現 75 分鐘，結束代碼 0。這段輔助碼不用學生背誦。

````python
import subprocess
import sys
from pathlib import Path

script = Path("worklog.py").resolve()
if not script.is_file():
    print("請先執行『產生檔案』那一格。")
else:
    result = subprocess.run(
        [sys.executable, "-X", "utf8", str(script)],
        capture_output=True, text=True, encoding="utf-8", timeout=10
    )
    print(result.stdout, end="")
    if result.stderr:
        print(result.stderr)
    print("結束代碼：", result.returncode, "（0 代表成功）")
    if result.returncode != 0:
        raise RuntimeError("執行失敗，請保留錯誤訊息。")
````

### 驗證正確與錯誤輸入
**這格在做什麼**：自行載入 `.py` 裡的函式，檢查 8 組已知答案。預期看到「通過 8 組檢查」。這格不需要先跑其他格子的 import 或函式定義，只需已存在 worklog.py。

| 測試資料 | 預期 |
|---|---|
| 45、30 | 75 |
| 空清單 | 0 |
| 0 | 0 |
| 45、30、20 | 95 |
| -1 | 拒絕負數 |
| 文字 "30" | 拒絕文字 |
| 1.5 | 拒絕小數 |
| True | 拒絕布林值 |

````python
import importlib.util
from pathlib import Path

script = Path("worklog.py").resolve()
if not script.is_file():
    print("請先執行『產生檔案』那一格。")
else:
    spec = importlib.util.spec_from_file_location("lesson_worklog", script)
    module = importlib.util.module_from_spec(spec)
    spec.loader.exec_module(module)
    for values, expected in [([45, 30], 75), ([], 0), ([0], 0), ([45, 30, 20], 95)]:
        assert module.total_minutes(values) == expected
    for values, error in [([-1], ValueError), (["30"], TypeError), ([1.5], TypeError), ([True], TypeError)]:
        try:
            module.total_minutes(values)
        except error:
            pass
        else:
            raise AssertionError(f"這組資料應該被拒絕：{values!r}")
    print("通過 8 組檢查：4 組正確資料，以及 4 組不合法輸入。")
````
